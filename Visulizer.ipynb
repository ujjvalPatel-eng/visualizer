```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import os


class SalesData:
    def __init__(self):
        self.__data = None

    def load_data(self, file_path):
        try:
            self.__data = pd.read_csv(file_path)
            print("File loaded successfully.")
        except Exception as e:
            print(f"Error reading the file: {e}")

    def display(self):
        try:
            if self.__data is not None:
                print("Summary of Sales Data:")
                print(self.__data)
            else:
                print("No data to display.")
        except Exception as e:
            print(f"Error displaying data: {e}")

    def handle_missing_values(self):

        if self.__data is None:
            print("No data loaded.")
            return

        while True:

            print("\n" + "-" * 20)
            print("Handle Missing Values")
            print("-" * 20)

            print("1. Show Missing Values Count")
            print("2. Drop Rows with Missing Values")
            print("3. Fill Missing Values")
            print("4. Drop Columns with Missing Values")
            print("5. Exit")

            choice = input("\nEnter your choice: ")

            match choice:

                case "1":

                    print("\nMissing Values Count:")
                    print(self.__data.isnull().sum())

                case "2":

                    confirm = input(
                        "Are you sure you want to drop rows with missing values? (yes/no): "
                    ).lower()

                    if confirm == "yes":

                        before = self.__data.shape

                        self.__data.dropna(inplace=True)

                        after = self.__data.shape

                        print("\nRows with missing values removed.")
                        print(f"Before Shape: {before}")
                        print(f"After Shape: {after}")

                case "3":

                    print("\nAvailable Columns:")
                    print(self.__data.columns.tolist())

                    column = input("Enter column name: ")

                    print("\nFill Methods:")
                    print("1. Mean")
                    print("2. Median")
                    print("3. Mode")
                    print("4. Custom Value")

                    method = input("Choose fill method: ")

                    try:

                        if method == "1":

                            self.__data[column].fillna(
                                self.__data[column].mean(), inplace=True
                            )

                        elif method == "2":

                            self.__data[column].fillna(
                                self.__data[column].median(), inplace=True
                            )

                        elif method == "3":

                            self.__data[column].fillna(
                                self.__data[column].mode()[0], inplace=True
                            )

                        elif method == "4":

                            value = input("Enter custom value: ")

                            self.__data[column].fillna(value, inplace=True)

                        else:
                            print("Invalid method.")
                            continue

                        print("\nMissing values filled successfully.")

                    except Exception as e:
                        print(f"Error: {e}")

                case "4":

                    confirm = input(
                        "Are you sure you want to drop columns with missing values? (yes/no): "
                    ).lower()

                    if confirm == "yes":

                        before = self.__data.shape

                        self.__data.dropna(axis=1, inplace=True)

                        after = self.__data.shape

                        print("\nColumns with missing values removed.")
                        print(f"Before Shape: {before}")
                        print(f"After Shape: {after}")

                case "5":

                    print("Exiting Missing Values Menu.")
                    break

                case _:

                    print("Invalid choice.")

    def explore_data(self):
        print("-" * 15)
        print(" Explore data:")
        print("-" * 15)

        if self.__data is not None:
            while True:
                try:
                    print("\nOptions:")
                    print("\n1. First 10 Rows")
                    print("2. Last 10 Rows")
                    print("3. Column Names")
                    print("4. Basic Information")
                    print("5. Statistical Summary")
                    print("6. Exit")

                    choice = input("Enter your choice (1-6): " )                    
                    if (choice is None):
                        print("Exiting the program.")
                        break

                    if choice == "1":
                        print("\nFirst 10 Rows:")
                        print(self.__data.head(10))
                    elif choice == "2":
                        print("\nLast 10 Rows:")
                        print(self.__data.tail(10))
                    elif choice == "3":
                        print("\nColumn Names:")
                        print(self.__data.columns.tolist())
                    elif choice == "4":
                        print("\nBasic Information:")
                        print(self.__data.info())
                    elif choice == "5":
                        print("\nStatistical Summary:")
                        print(self.__data.describe())
                    else:
                        print(
                            "Invalid choice. Please enter a valid number "
                        )
                except Exception as e:
                    print(f"An error occurred: {e}")

    def dataframe_operations(self):

        if self.__data is None:
            print("No data loaded.")
            return

        while True:

            print("\n" + "-" * 20)
            print("DataFrame Operations")
            print("-" * 20)

            print("1. Display Shape")
            print("2. Select Columns")
            print("3. Filter Rows")
            print("4. Sort Values")
            print("5. Group By")
            print("6. Find Maximum Value")
            print("7. Find Minimum Value")
            print("8. Add New Column")
            print("9. Exit")

            choice = input("\nEnter your choice: ")

            match choice:

                case "1":
                    print("\nShape of DataFrame:")
                    print(self.__data.shape)

                case "2":
                    print("\nAvailable Columns:")
                    print(self.__data.columns.tolist())
                    cols = input("Enter column names separated by commas: ").split(",")
                    cols = [col.strip() for col in cols]

                    try:
                        print(self.__data[cols])
                    except Exception as e:
                        print(f"Error: {e}")

                case "3":
                    print("\nAvailable Columns:")
                    print(self.__data.columns.tolist())

                    column = input("Enter column name for filtering: ")

                    value = input("Enter value to filter: ")

                    try:
                        print(self.__data[self.__data[column] == value])
                    except Exception as e:
                        print(f"Error: {e}")

                case "4":
                    print("\nAvailable Columns:")
                    print(self.__data.columns.tolist())

                    column = input("Enter column name to sort by: ")

                    order = input("Ascending? (yes/no): ").lower()

                    ascending = True if order == "yes" else False

                    try:
                        print(self.__data.sort_values(by=column, ascending=ascending))
                    except Exception as e:
                        print(f"Error: {e}")

                case "5":
                    print("\nAvailable Columns:")
                    print(self.__data.columns.tolist())
                    group_cols = input(
                        "Enter column names for groupby separated by commas: "
                    ).split(",")

                    group_cols = [col.strip() for col in group_cols]
                    agg_col = input("Enter numeric column for aggregation: ")
                    operation = input(
                        "Choose aggregation (sum, mean, max, min, count): "
                    ).lower()

                    try:

                        if operation == "sum":
                            result = self.__data.groupby(group_cols)[agg_col].sum()

                        elif operation == "mean":
                            result = self.__data.groupby(group_cols)[agg_col].mean()

                        elif operation == "max":
                            result = self.__data.groupby(group_cols)[agg_col].max()

                        elif operation == "min":
                            result = self.__data.groupby(group_cols)[agg_col].min()

                        elif operation == "count":
                            result = self.__data.groupby(group_cols)[agg_col].count()

                        else:
                            print("Invalid aggregation operation.")
                            continue

                        print("\nGrouped Result:")
                        print(result)

                    except Exception as e:
                        print(f"Error: {e}")
                case "6":
                    print("\nAvailable Columns:")
                    print(self.__data.columns.tolist())

                    column = input("Enter numeric column name: ")

                    try:
                        print("\nMaximum Value:")
                        print(self.__data[column].max())
                    except Exception as e:
                        print(f"Error: {e}")

                case "7":
                    print("\nAvailable Columns:")
                    print(self.__data.columns.tolist())

                    column = input("Enter numeric column name: ")

                    try:
                        print("\nMinimum Value:")
                        print(self.__data[column].min())
                    except Exception as e:
                        print(f"Error: {e}")

                case "8":
                    new_col = input("Enter new column name: ")

                    value = input("Enter value for all rows: ")

                    self.__data[new_col] = value

                    print("\nNew column added successfully.")
                    print(self.__data.head())

                case "9":
                    print("Exiting DataFrame Operations.")
                    break

                case _:
                    print("Invalid choice.")
    def save_plot(self):

        directory = "saved_plots"
        os.makedirs(directory, exist_ok=True)

        choice = input(
            "\nDo you want to save this plot? (yes/no): "
        ).lower()
    
        if choice == "yes":
        
            filename = input("Enter filename: ")
    
            plt.savefig(os.path.join(directory, f"{filename}.png"))
    
            print(f"Plot saved as {directory}/{filename}.png")
    
        plt.close()
    def visualize_data(self):
        
        if self.__data is None:
            print("No data loaded.")
            return

        while True:

            print("\n" + "-" * 20)
            print("Data Visualization")
            print("-" * 20)
    
            print("1. Line Plot")
            print("2. Bar Plot")
            print("3. Histogram")
            print("4. Scatter Plot")
            print("5. Box Plot")
            print("6. Heatmap")
            print("7. Count Plot")
            print("8. Pie Chart")
            print("9. Exit")
    
            choice = input("\nEnter your choice: ")
    
            try:

                match choice:

                    case "1":

                        print("\nAvailable Columns:")
                        print(self.__data.columns.tolist())

                        x = input("Enter X-axis column: ")
                        y = input("Enter Y-axis column: ")

                        plt.figure(figsize=(8,5))

                        sns.lineplot(data=self.__data, x=x, y=y)

                        plt.title(f"{y} vs {x}")
                        plt.show()
                        self.save_plot()

                    case "2":

                        print("\nAvailable Columns:")
                        print(self.__data.columns.tolist())

                        x = input("Enter categorical column: ")
                        y = input("Enter numeric column: ")

                        plt.figure(figsize=(8,5))

                        sns.barplot(data=self.__data, x=x, y=y)

                        plt.title(f"{y} by {x}")
                        plt.xticks(rotation=45)

                        plt.show()
                        self.save_plot()

                    case "3":

                        print("\nAvailable Columns:")
                        print(self.__data.columns.tolist())

                        column = input("Enter numeric column: ")

                        plt.figure(figsize=(8,5))

                        sns.histplot(self.__data[column], kde=True)

                        plt.title(f"Histogram of {column}")

                        plt.show()
                        self.save_plot()

                    case "4":

                        print("\nAvailable Columns:")
                        print(self.__data.columns.tolist())

                        x = input("Enter X-axis column: ")
                        y = input("Enter Y-axis column: ")

                        plt.figure(figsize=(8,5))

                        sns.scatterplot(data=self.__data, x=x, y=y)

                        plt.title(f"{y} vs {x}")

                        plt.show()
                        self.save_plot()
                    case "5":

                        print("\nAvailable Columns:")
                        print(self.__data.columns.tolist())

                        column = input("Enter numeric column: ")

                        plt.figure(figsize=(8,5))

                        sns.boxplot(y=self.__data[column])

                        plt.title(f"Box Plot of {column}")

                        plt.show()
                        self.save_plot()

                    case "6":

                        plt.figure(figsize=(10,6))

                        numeric_data = self.__data.select_dtypes(
                            include=['number']
                        )

                        sns.heatmap(
                            numeric_data.corr(),
                            annot=True,
                            cmap="coolwarm"
                        )

                        plt.title("Correlation Heatmap")

                        plt.show()
                        self.save_plot()

                    case "7":

                        print("\nAvailable Columns:")
                        print(self.__data.columns.tolist())

                        column = input("Enter categorical column: ")

                        plt.figure(figsize=(8,5))

                        sns.countplot(data=self.__data, x=column)

                        plt.title(f"Count Plot of {column}")

                        plt.xticks(rotation=45)

                        plt.show()
                        self.save_plot()

                    case "8":

                        print("\nAvailable Columns:")
                        print(self.__data.columns.tolist())

                        column = input("Enter categorical column: ")

                        counts = self.__data[column].value_counts()

                        plt.figure(figsize=(7,7))

                        plt.pie(
                            counts,
                            labels=counts.index,
                            autopct='%1.1f%%'
                        )

                        plt.title(f"Pie Chart of {column}")

                        plt.show()
                        self.save_plot()

                    case "9":

                        print("Exiting Visualization Menu.")
                        break

                    case _:

                        print("Invalid choice.")

            except Exception as e:
                print(f"Error: {e}")
    

print("-" * 20)
print("Welcome to the Sales Data Explorer!")
print("-" * 20)
object = SalesData()
while True:
    print("\nOptions:")
    print("\n1. Load Sales Data")
    print("2. Explore data")
    print("3. perform dataframe operations".title())
    print("4.handle missing values".title())
    print("5.visualize data".title())
    print("6. Exit")

    match input("Enter your choice (1-6) or press Enter / type 'q'/'esc' to quit: "):
        case "1":
            # Load Sales Data
            file_path = input("Enter the file path of the sales data (CSV format): ")
            object.load_data(file_path)

        case "2":
            object.explore_data()
        case "3":
            object.dataframe_operations()
        case "4":
            object.handle_missing_values()
        case "5":
            object.visualize_data()
        case "6":
            break
        case _:
            print(
                "Invalid choice. Please enter a number between 1 and 6, or 'q' to quit."
            )
```
